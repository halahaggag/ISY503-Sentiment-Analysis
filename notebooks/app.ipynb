```python
%%writefile app.py

import streamlit as st
import tensorflow as tf
import pickle
import re
from tensorflow.keras.preprocessing.sequence import pad_sequences

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

MODEL_PATH = "sentiment_model.keras"
TOKENIZER_PATH = "sentiment_tokenizer.pkl"

MAX_LENGTH = 100


# ------------------------------------------------------------
# Load model and tokenizer
# ------------------------------------------------------------

model = tf.keras.models.load_model(MODEL_PATH)

with open(TOKENIZER_PATH, "rb") as file:
    tokenizer = pickle.load(file)


# ------------------------------------------------------------
# Text cleaning function
# ------------------------------------------------------------

def clean_review(text):

    text = text.lower()

    # Remove HTML tags
    text = re.sub(r"<.*?>", " ", text)

    # Keep letters and apostrophes
    text = re.sub(r"[^a-zA-Z']", " ", text)

    # Remove repeated spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text


# ------------------------------------------------------------
# Prediction function
# ------------------------------------------------------------

def predict_sentiment(text):

    cleaned_text = clean_review(text)

    sequence = tokenizer.texts_to_sequences(
        [cleaned_text]
    )

    padded_sequence = pad_sequences(
        sequence,
        maxlen=MAX_LENGTH,
        padding="post",
        truncating="post"
    )

    probability = model.predict(
        padded_sequence,
        verbose=0
    )[0][0]

    if probability >= 0.5:
        sentiment = "Positive review"
    else:
        sentiment = "Negative review"

    return sentiment, float(probability)


# ------------------------------------------------------------
# Streamlit interface
# ------------------------------------------------------------

st.title("Amazon Review Sentiment Analysis")

st.write(
    "Enter a product review below and the trained neural network "
    "will classify it as positive or negative."
)

user_review = st.text_area(
    "Enter review:",
    height=150
)

if st.button("Analyse Sentiment"):

    if user_review.strip() == "":
        st.warning("Please enter a review.")

    else:
        sentiment, probability = predict_sentiment(
            user_review
        )

        st.subheader("Prediction")

        st.write(sentiment)

        st.write(
            f"Positive sentiment probability: "
            f"{probability:.2%}"
        )
```
